```python
# --- importacion de librerias ---#

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# --- cargar archivos ---#

df = pd.read_csv('GoogleAds_DataAnalytics_Sales_Uncleaned.csv')
```

```python
# --- exploracion de datos ---#

df.head(n=10)
```

```python
df.tail(n=10)
```

```python
# --- eliminar datos duplicados ---#

df.duplicated().sum()
```

```python
# --- limpiar datos de tipo str ---#
df['Campaign_Name'].value_counts()

df['Campaign_Name'] = df['Campaign_Name'].replace({
    'Data Analytics Corse':'Data Analytcis Course',
    'DataAnalyticsCourse': 'Data Analytcis Course',
    'Data Anlytics Corse': 'Data Analytcis Course'
    })

df['Campaign_Name'].value_counts()
```

```python
df['Location'].value_counts()

df['Location'] = df['Location'].replace({
    'HYDERABAD':'Hyderabad',
    'hyderabad': 'Hyderabad',
    'Hyderbad': 'Hyderabad',
    'hydrebad': 'Hyderabad'
    })

df['Location'].value_counts()
```

```python
df['Device'].value_counts()

df['Device'] = df['Device'].replace({
    'MOBILE':'Mobile',
    'mobile': 'Mobile',
    'desktop':'Desktop',
    'DESKTOP':'Desktop',
    'TABLET':'Tablet',
    'tablet':'Tablet'
    })

df['Device'].value_counts()
```

```python
df['Keyword'].value_counts()

df['Keyword'] = df['Keyword'].replace({
    'online data analytic':'data analytics online',
    'data analitics online':'data analytics online',
    'data anaytics training':'data analytics training'
})

df['Keyword'].value_counts()
```

```python
#--- buscar datos nulos ---#

df.isna().sum()
```

```python
# esta columna es mejor recrearla en sql
df = df.drop(columns=['Conversion Rate'])

df.columns
```

```python
df.head()
```

```python
mean_clicks = df['Clicks'].median()
df['Clicks'] = df['Clicks'].fillna(mean_clicks).round().astype(int)

df['Clicks'].tail(n=10)
```

```python
mean_clicks = df['Impressions'].median()
df['Impressions'] = df['Impressions'].fillna(mean_clicks).round().astype(int)
```

```python
df.isna().sum()
```

```python
df['Cost'] = (
    df['Cost'].str.replace("$","", regex=False)
    )

mean_cost = df['Cost'].median()
df['Cost'] = df['Cost'].fillna(mean_cost).astype(float)
```

```python
df.isna().sum()
```

```python
mean_leads = df['Leads'].median()
df['Leads'] = df['Leads'].fillna(mean_leads).round().astype(int)
```

```python
mean_conversions = df['Conversions'].median()
df['Conversions'] = df['Conversions'].fillna(mean_conversions).round().astype(int)
```

```python
df['Sale_Amount'] = (
    df['Sale_Amount'].str.replace("$","", regex=False)
    )

mean_sale_amount = df['Sale_Amount'].median()
df['Sale_Amount'] = df['Sale_Amount'].fillna(mean_sale_amount).astype(float)
```

```python
df.isna().sum()
```

```python
df['Ad_Date'] = df['Ad_Date'].astype(str)

df['Ad_Date'] = df['Ad_Date'].str.replace('/','-', regex = False)

df['Ad_Date'] = pd.to_datetime(df['Ad_Date'], dayfirst= True, errors='coerce')
```

```python
# --- exportar a sqlite ---#

import sqlite3

db_clean = "google_ads_sales_clean.db"

conn = sqlite3.connect(db_clean)

df.to_sql('ads_data_clean', conn, if_exists='replace', index=False)

conn.close()
```
